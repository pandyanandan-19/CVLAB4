```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def show_comparision(oringinal,robot,previtt,canny):
  plt.figure(figsize=(12,10))

  plt.subplot(2,2,1)
  plt.imshow(oringinal,cmap='gray')
  plt.title('Original grayscale image')
  plt.axis('off')

  plt.subplot(2,2,2)
  plt.imshow(robot,cmap='gray')
  plt.title('Robot edge detection')
  plt.axis('off')

  plt.subplot(2,2,3)
  plt.imshow(previtt,cmap='gray')
  plt.title('Prewitt edge detection')
  plt.axis('off')

  plt.subplot(2,2,4)
  plt.imshow(canny,cmap='gray')
  plt.title('Canny edge detection')
  plt.axis('off')

  plt.tight_layout()
  plt.show()
```

```python
img = cv2.imread('/content/drive/MyDrive/ICHI1.jpg',0)
if img is None:
  print('Error loading image')
else:
  soblex = cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
  sobley = cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)
  kernelx = np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
  kernely = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
  prewittx = cv2.filter2D(img, -1, kernelx)
  prewitty = cv2.filter2D(img, -1, kernely)
  prewitt_com = prewittx + prewitty
  canny_edges = cv2.Canny(img, 100, 200)
  show_comparision(img,soblex,sobley,canny_edges)
```
